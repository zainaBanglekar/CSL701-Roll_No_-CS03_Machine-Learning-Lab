## Relevant Theory

### Logistic Regression

Logistic Regression is a supervised machine learning algorithm used for **binary classification**. In this experiment, it is used to classify credit card transactions into two classes:

- **0 $\rightarrow$ Legitimate transaction**
- **1 $\rightarrow$ Fraudulent transaction**

The model first calculates a linear combination of input features:

$$
z = w^T X + b
$$

where:

- $w$ = weight vector
- $X$ = input feature vector
- $b$ = bias/intercept
- $z$ = linear predictor

---

### Sigmoid Function

The Sigmoid Function converts the value of $z$ into a probability between 0 and 1:

$$
\sigma(z) = \frac{1}{1 + e^{-z}}
$$

The predicted probability that a transaction is fraudulent is:

$$
P(y=1|X) = \sigma(w^T X + b)
$$

The predicted class can be obtained using a threshold:

$$
\hat{y} =
\begin{cases}
1, & \text{if } P(y=1|X) \geq 0.5 \\
0, & \text{if } P(y=1|X) < 0.5
\end{cases}
$$

---
## Cost Function

Logistic Regression uses the **Binary Cross-Entropy Loss** or **Log Loss**.

The cost function is:

$$
J(w,b) =
-\frac{1}{m}
\sum_{i=1}^{m}
\left[
y_i \log(\hat{y}_i)
+
(1-y_i)\log(1-\hat{y}_i)
\right]
$$

where:

- $m$ = number of training examples
- $y_i$ = actual class label
- $\hat{y}_i$ = predicted probability

The objective of training is to find values of $w$ and $b$ that minimize the cost function.

---

## Gradient Descent

Gradient Descent is used to optimize the parameters of the Logistic Regression model.

The weights are updated using:

$$
w := w - \alpha \frac{\partial J}{\partial w}
$$

The bias is updated using:

$$
b := b - \alpha \frac{\partial J}{\partial b}
$$

where $\alpha$ is the learning rate.

For Logistic Regression, the gradient of the cost function with respect to the weights can be expressed as:

$$
\frac{\partial J}{\partial w}
=
\frac{1}{m}X^T(\hat{y}-y)
$$

The gradient with respect to the bias is:

$$
\frac{\partial J}{\partial b}
=
\frac{1}{m}
\sum_{i=1}^{m}
(\hat{y}_i-y_i)
$$

Therefore, the parameter updates become:

$$
w := w - \alpha
\left[
\frac{1}{m}X^T(\hat{y}-y)
\right]
$$

$$
b := b - \alpha
\left[
\frac{1}{m}
\sum_{i=1}^{m}
(\hat{y}_i-y_i)
\right]
$$

---

## Class Imbalance

Credit card fraud datasets are generally highly imbalanced because legitimate transactions significantly outnumber fraudulent transactions.

For example, the dataset may contain a very large number of legitimate transactions and only a small number of fraudulent transactions.

A model that predicts most transactions as legitimate may achieve high accuracy but perform poorly at detecting fraud.

Therefore, techniques such as:

- Random Undersampling
- Oversampling
- Class Weighting

can be used to handle class imbalance.

## Evaluation Metrics

### Confusion Matrix

The confusion matrix consists of four components:

- **True Positive (TP):** Fraudulent transaction correctly identified as fraud.
- **True Negative (TN):** Legitimate transaction correctly identified as legitimate.
- **False Positive (FP):** Legitimate transaction incorrectly identified as fraud.
- **False Negative (FN):** Fraudulent transaction incorrectly identified as legitimate.

---

### Accuracy

Accuracy measures the proportion of correctly classified transactions:

$$
Accuracy =
\frac{TP+TN}
{TP+TN+FP+FN}
$$

However, accuracy alone is not sufficient for highly imbalanced fraud detection datasets.

---

### Precision

Precision indicates the proportion of transactions predicted as fraudulent that are actually fraudulent:

$$
Precision =
\frac{TP}
{TP+FP}
$$

A high Precision means that the model produces fewer false fraud alerts.

---

### Recall

Recall measures the proportion of actual fraudulent transactions correctly detected:

$$
Recall =
\frac{TP}
{TP+FN}
$$

Recall is particularly important in fraud detection because a **False Negative** represents a fraudulent transaction that was not detected.

---

### F1-Score

F1-Score is the harmonic mean of Precision and Recall:

$$
F1 =
2 \times
\frac{Precision \times Recall}
{Precision + Recall}
$$

---

## ROC Curve

The Receiver Operating Characteristic (ROC) curve represents the relationship between the **True Positive Rate (TPR)** and **False Positive Rate (FPR)** at different classification thresholds.

The True Positive Rate is:

$$
TPR =
\frac{TP}
{TP+FN}
$$

The False Positive Rate is:

$$
FPR =
\frac{FP}
{FP+TN}
$$

The ROC curve helps determine how well the model separates fraudulent transactions from legitimate transactions.

---

## *Area Under the Curve (AUC)*

The **Area Under the ROC Curve (AUC)** summarizes the performance of the classifier across different classification thresholds.

An AUC value close to:

$$
AUC = 1
$$

indicates excellent discrimination, while:

$$
AUC = 0.5
$$

indicates performance approximately equivalent to random classification.

```python
# ==========================================
# TASK 3: DATA PREPROCESSING
# ==========================================

# ------------------------------------------
# 1. Check Missing Values
# ------------------------------------------

print("========== MISSING VALUES ==========")

missing_values = df.isnull().sum()

print(missing_values)

print("\nTotal Missing Values:",
      missing_values.sum())


# ------------------------------------------
# 2. Handle Missing Values
# ------------------------------------------

if missing_values.sum() == 0:
    print("\nNo missing values found in the dataset.")
else:
    print("\nMissing values found.")

    # Separate numerical and categorical columns
    numerical_columns = df.select_dtypes(
        include=np.number
    ).columns

    categorical_columns = df.select_dtypes(
        exclude=np.number
    ).columns

    # Fill numerical missing values with median
    for column in numerical_columns:
        if df[column].isnull().sum() > 0:
            df[column] = df[column].fillna(
                df[column].median()
            )

    # Fill categorical missing values with mode
    for column in categorical_columns:
        if df[column].isnull().sum() > 0:
            df[column] = df[column].fillna(
                df[column].mode()[0]
            )

    print("Missing values handled successfully.")


# ------------------------------------------
# 3. Check Duplicate Records
# ------------------------------------------

print("\n========== DUPLICATE RECORDS ==========")

duplicate_count = df.duplicated().sum()

print("Number of duplicate records:",
      duplicate_count)


# ------------------------------------------
# 4. Remove Duplicate Records
# ------------------------------------------

if duplicate_count > 0:
    df = df.drop_duplicates()

    print("Duplicate records removed.")
else:
    print("No duplicate records found.")


print("\nDataset shape after removing duplicates:")
print(df.shape)


# ------------------------------------------
# 5. Identify Categorical Variables
# ------------------------------------------

print("\n========== CATEGORICAL VARIABLES ==========")

categorical_columns = df.select_dtypes(
    exclude=np.number
).columns.tolist()

if len(categorical_columns) == 0:
    print("No categorical variables found.")
else:
    print(categorical_columns)


# ------------------------------------------
# 6. Identify Numerical Variables
# ------------------------------------------

print("\n========== NUMERICAL VARIABLES ==========")

numerical_columns = df.select_dtypes(
    include=np.number
).columns.tolist()

print(numerical_columns)


# ------------------------------------------
# 7. Separate Target Variable
# ------------------------------------------

y = df["fraud"]

X = df.drop("fraud", axis=1)


# ------------------------------------------
# 8. One-Hot Encoding for Categorical
#    Variables
# ------------------------------------------

if len(categorical_columns) > 0:

    X = pd.get_dummies(
        X,
        columns=categorical_columns,
        drop_first=True
    )

    print("\nCategorical variables converted using")
    print("One-Hot Encoding.")

else:

    print("\nNo categorical variables require encoding.")


# ------------------------------------------
# 9. Convert Boolean Columns to Integer
# ------------------------------------------

boolean_columns = X.select_dtypes(
    include="bool"
).columns

if len(boolean_columns) > 0:
    X[boolean_columns] = X[boolean_columns].astype(int)


# ------------------------------------------
# 10. Identify Final Numerical Features
# ------------------------------------------

print("\n========== FINAL FEATURES ==========")

print("Number of features:",
      X.shape[1])

print("\nFeatures used for Logistic Regression:")

print(X.columns.tolist())


# ------------------------------------------
# 11. Check Final Missing Values
# ------------------------------------------

print("\n========== FINAL MISSING VALUE CHECK ==========")

print(X.isnull().sum().sum())

print("Total missing values in feature dataset:",
      X.isnull().sum().sum())


# ------------------------------------------
# 12. Check Data Types
# ------------------------------------------

print("\n========== FINAL DATA TYPES ==========")

print(X.dtypes)


# ------------------------------------------
# 13. Final Dataset Summary
# ------------------------------------------

print("\n========== PREPROCESSING SUMMARY ==========")

print("Dataset shape:", df.shape)

print("Feature matrix shape:", X.shape)

print("Target variable:", y.name)

print("Number of numerical features:",
      len(X.select_dtypes(include=np.number).columns))

print("Number of categorical features:",
      len(X.select_dtypes(exclude=np.number).columns))

print("Missing values:",
      X.isnull().sum().sum())

print("Duplicate records:",
      df.duplicated().sum())


# ------------------------------------------
# 14. Display Final Feature Dataset
# ------------------------------------------

print("\n========== FINAL FEATURE DATA ==========")

display(X.head())

print("\n========== TARGET DATA ==========")

display(y.head())
```

## Task 4: Feature Scaling and Class Imbalance Handling

Standardize the numerical features using **Standardization** so that they have approximately zero mean and unit variance.

The standardized value of a feature is calculated as:

$$
z = \frac{x-\mu}{\sigma}
$$

where:

* (x) = original feature value
* (\mu) = mean of the feature
* (\sigma) = standard deviation of the feature
* (z) = standardized feature value

Handle the **class imbalance** in the dataset using a suitable technique such as:

* Random Undersampling
* Oversampling
* Class Weighting

Display the class distribution **before and after** applying the selected class-imbalance handling technique.

Compare the number of legitimate and fraudulent transactions before and after balancing.

The objective of this task is to ensure that the features are on a comparable scale and that the Logistic Regression model does not become biased toward the majority class.


```python
# ==========================================
# TASK 4: FEATURE SCALING AND
#          CLASS IMBALANCE HANDLING
# ==========================================

# Import required libraries
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from imblearn.over_sampling import RandomOverSampler


# ------------------------------------------
# 1. Separate Features and Target
# ------------------------------------------

X = df.drop("fraud", axis=1)
y = df["fraud"]

print("Feature Shape:", X.shape)
print("Target Shape:", y.shape)


# ------------------------------------------
# 2. Display Original Class Distribution
# ------------------------------------------

print("\n========== CLASS DISTRIBUTION BEFORE BALANCING ==========")

before_distribution = y.value_counts().sort_index()

print("Legitimate transactions (0):",
      before_distribution.get(0, 0))

print("Fraudulent transactions (1):",
      before_distribution.get(1, 0))


# Calculate percentages
before_percentage = y.value_counts(
    normalize=True
).sort_index() * 100

print("\nClass Percentage Before Balancing:")

print("Legitimate (0): {:.2f}%".format(
    before_percentage.get(0, 0)
))

print("Fraudulent (1): {:.2f}%".format(
    before_percentage.get(1, 0)
))


# ------------------------------------------
# 3. Train-Test Split
# ------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\n========== TRAIN-TEST SPLIT ==========")

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)


# ------------------------------------------
# 4. Feature Scaling
# ------------------------------------------

scaler = StandardScaler()

# Fit only on training data
X_train_scaled = scaler.fit_transform(X_train)

# Transform testing data
X_test_scaled = scaler.transform(X_test)

print("\n========== FEATURE SCALING ==========")
print("Standardization completed successfully.")

print("\nMean of scaled training features:")
print(np.mean(X_train_scaled, axis=0))

print("\nStandard deviation of scaled training features:")
print(np.std(X_train_scaled, axis=0))


# ------------------------------------------
# 5. Random Oversampling
# ------------------------------------------

ros = RandomOverSampler(
    random_state=42
)

X_train_balanced, y_train_balanced = ros.fit_resample(
    X_train_scaled,
    y_train
)


# ------------------------------------------
# 6. Class Distribution After Balancing
# ------------------------------------------

print("\n========== CLASS DISTRIBUTION AFTER BALANCING ==========")

after_distribution = y_train_balanced.value_counts(
).sort_index()

print("Legitimate transactions (0):",
      after_distribution.get(0, 0))

print("Fraudulent transactions (1):",
      after_distribution.get(1, 0))


# Calculate percentages after balancing
after_percentage = y_train_balanced.value_counts(
    normalize=True
).sort_index() * 100

print("\nClass Percentage After Balancing:")

print("Legitimate (0): {:.2f}%".format(
    after_percentage.get(0, 0)
))

print("Fraudulent (1): {:.2f}%".format(
    after_percentage.get(1, 0)
))


# ------------------------------------------
# 7. Compare Before and After Balancing
# ------------------------------------------

comparison = pd.DataFrame({
    "Class": [
        "Legitimate (0)",
        "Fraudulent (1)"
    ],

    "Before Balancing": [
        y_train.value_counts().get(0, 0),
        y_train.value_counts().get(1, 0)
    ],

    "After Balancing": [
        y_train_balanced.value_counts().get(0, 0),
        y_train_balanced.value_counts().get(1, 0)
    ]
})

print("\n========== BEFORE VS AFTER BALANCING ==========")

display(comparison)


# ------------------------------------------
# 8. Plot Class Distribution Before
#    and After Balancing
# ------------------------------------------

plt.figure(figsize=(10, 5))

classes = ["Legitimate (0)", "Fraudulent (1)"]

before_counts = [
    y_train.value_counts().get(0, 0),
    y_train.value_counts().get(1, 0)
]

after_counts = [
    y_train_balanced.value_counts().get(0, 0),
    y_train_balanced.value_counts().get(1, 0)
]

x = np.arange(len(classes))
width = 0.35

plt.bar(
    x - width / 2,
    before_counts,
    width,
    label="Before Balancing"
)

plt.bar(
    x + width / 2,
    after_counts,
    width,
    label="After Balancing"
)

plt.xlabel("Transaction Class")
plt.ylabel("Number of Transactions")

plt.title(
    "Class Distribution Before and After Random Oversampling"
)

plt.xticks(x, classes)

plt.legend()

plt.show()


# ------------------------------------------
# 9. Final Shapes
# ------------------------------------------

print("\n========== FINAL DATA SHAPES ==========")

print("Original X:", X.shape)
print("Original y:", y.shape)

print("\nScaled Training Data:",
      X_train_scaled.shape)

print("Scaled Testing Data:",
      X_test_scaled.shape)

print("\nBalanced Training Data:",
      X_train_balanced.shape)

print("Balanced Target Data:",
      y_train_balanced.shape)


# ------------------------------------------
# 10. Task 4 Summary
# ------------------------------------------

print("\n========== TASK 4 SUMMARY ==========")

print("Feature scaling method: Standardization")

print("Class imbalance technique: Random Oversampling")

print("\nTraining class distribution before balancing:")
print(y_train.value_counts())

print("\nTraining class distribution after balancing:")
print(y_train_balanced.value_counts())

print("\nTask 4 completed successfully!")
```

## Task 6: Implement Sigmoid and Cost Functions

Implement the **Sigmoid Function** used in Logistic Regression.

The Sigmoid Function is defined as:

$$
\sigma(z) = \frac{1}{1+e^{-z}}
$$

The Sigmoid Function converts the output of the linear equation into a probability value between 0 and 1.

Implement the **Binary Cross-Entropy Cost Function**:

$$
J(w,b) =
-\frac{1}{m}
\sum_{i=1}^{m}
\left[
y_i\log(\hat{y}_i)
+
(1-y_i)\log(1-\hat{y}_i)
\right]
$$

where:

* (m) = number of training samples
* (y_i) = actual class label
* (\hat{y}_i) = predicted probability
* (w) = weight vector
* (b) = bias

Test the Sigmoid Function using suitable sample values.

Calculate the initial cost of the Logistic Regression model using the initialized weights and bias.

The objective of this task is to understand how the Sigmoid Function produces probabilities and how the Binary Cross-Entropy Cost Function measures the error of the Logistic Regression model.


```python
# ==========================================
# TASK 6: SIGMOID AND COST FUNCTIONS
# ==========================================

import numpy as np


# ------------------------------------------
# 1. Define Sigmoid Function
# ------------------------------------------

def sigmoid(z):
    """
    Sigmoid function:

            1
    σ(z) = -------
           1 + e^(-z)

    Converts any real-valued number into
    a probability between 0 and 1.
    """

    return 1 / (1 + np.exp(-z))


# ------------------------------------------
# 2. Test Sigmoid Function
# ------------------------------------------

print("========== SIGMOID FUNCTION TEST ==========")

sample_values = np.array([
    -10,
    -5,
    -2,
    0,
    2,
    5,
    10
])

sigmoid_values = sigmoid(sample_values)

for z, result in zip(sample_values, sigmoid_values):
    print(f"Sigmoid({z:>3}) = {result:.6f}")


# ------------------------------------------
# 3. Define Binary Cross-Entropy Cost
# ------------------------------------------

def compute_cost(X, y, w, b):
    """
    Binary Cross-Entropy Cost Function:

             1
    J = - -------- Σ [y log(y_hat)
             m
              + (1-y) log(1-y_hat)]

    X : Feature matrix
    y : Actual target values
    w : Weight vector
    b : Bias
    """

    m = len(y)

    # Calculate linear combination
    z = np.dot(X, w) + b

    # Calculate predicted probabilities
    y_hat = sigmoid(z)

    # Avoid log(0)
    epsilon = 1e-15

    y_hat = np.clip(
        y_hat,
        epsilon,
        1 - epsilon
    )

    # Calculate Binary Cross-Entropy Cost
    cost = -(1 / m) * np.sum(
        y * np.log(y_hat)
        +
        (1 - y) * np.log(1 - y_hat)
    )

    return cost


# ------------------------------------------
# 4. Prepare Training Data
# ------------------------------------------

# Convert training data to NumPy arrays
X_train_array = X_train.to_numpy()

y_train_array = y_train.to_numpy()


# ------------------------------------------
# 5. Standardize Training Features
# ------------------------------------------

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(
    X_train_array
)


# ------------------------------------------
# 6. Initialize Weights and Bias
# ------------------------------------------

# Number of features
n_features = X_train_scaled.shape[1]

# Initialize weights to zero
w_initial = np.zeros(n_features)

# Initialize bias to zero
b_initial = 0


print("\n========== INITIAL PARAMETERS ==========")

print("Number of features:", n_features)

print("Initial weights:")
print(w_initial)

print("Initial bias:")
print(b_initial)


# ------------------------------------------
# 7. Calculate Initial Cost
# ------------------------------------------

initial_cost = compute_cost(
    X_train_scaled,
    y_train_array,
    w_initial,
    b_initial
)


print("\n========== INITIAL COST ==========")

print("Initial Cost:",
      initial_cost)


# ------------------------------------------
# 8. Display Sample Predictions
# ------------------------------------------

z_sample = np.dot(
    X_train_scaled[:5],
    w_initial
) + b_initial

probability_sample = sigmoid(z_sample)


print("\n========== SAMPLE PREDICTIONS ==========")

for i in range(5):

    print(
        f"Sample {i + 1}: "
        f"z = {z_sample[i]:.4f}, "
        f"Probability = {probability_sample[i]:.4f}, "
        f"Actual = {y_train_array[i]}"
    )


# ------------------------------------------
# 9. Task 6 Summary
# ------------------------------------------

print("\n========== TASK 6 SUMMARY ==========")

print("Sigmoid function implemented successfully.")

print("Binary Cross-Entropy cost function implemented successfully.")

print("Weights initialized to zero.")

print("Bias initialized to zero.")

print(f"Initial Cost = {initial_cost:.6f}")

print("\nTask 6 completed successfully!")
```

## Task 7: Implement Logistic Regression Using Gradient Descent

Implement **Logistic Regression from scratch** using NumPy without directly using the Scikit-learn Logistic Regression implementation.

Initialize the model parameters, including the **weight vector** (w) and **bias** (b).

Calculate the predicted probability using the Sigmoid Function:

$$
\hat{y} = \sigma(Xw+b)
$$

where:

* (X) = input feature matrix
* (w) = weight vector
* $\hat{y} = \text{predicted probability}$


Calculate the gradients of the cost function with respect to the weights and bias:

<div align="left">

$$
\frac{\partial J}{\partial w}
=
\frac{1}{m}X^T(\hat{y}-y)
$$

$$
\frac{\partial J}{\partial b}
=
\frac{1}{m}\sum_{i=1}^{m}(\hat{y}_i-y_i)
$$

</div>

Update the model parameters using **Gradient Descent**:

$$
w := w-\alpha\frac{\partial J}{\partial w}
$$

$$
b := b-\alpha\frac{\partial J}{\partial b}
$$

where ($\alpha$) represents the **learning rate**.

Train the model for a suitable number of iterations and record the cost value during each iteration.

Plot **Cost vs. Iteration** to verify whether the optimization process is converging.

The objective of this task is to implement Logistic Regression manually and understand how Gradient Descent optimizes the model parameters by minimizing the Binary Cross-Entropy Cost Function.


```python
# ==========================================
# TASK 7: LOGISTIC REGRESSION USING
#          GRADIENT DESCENT
# ==========================================

import numpy as np
import matplotlib.pyplot as plt


# ------------------------------------------
# 1. Prepare Training Data
# ------------------------------------------

X_train_array = X_train.to_numpy()
y_train_array = y_train.to_numpy()

print("Training Data Shape:")
print("X_train:", X_train_array.shape)
print("y_train:", y_train_array.shape)


# ------------------------------------------
# 2. Feature Scaling
# ------------------------------------------

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(
    X_train_array
)

# Scale validation and test data using
# the same scaler fitted on training data

X_val_scaled = scaler.transform(
    X_val.to_numpy()
)

X_test_scaled = scaler.transform(
    X_test.to_numpy()
)


print("\nFeature scaling completed successfully.")


# ------------------------------------------
# 3. Define Sigmoid Function
# ------------------------------------------

def sigmoid(z):

    # Prevent overflow in exp()
    z = np.clip(z, -500, 500)

    return 1 / (1 + np.exp(-z))


# ------------------------------------------
# 4. Define Cost Function
# ------------------------------------------

def compute_cost(X, y, w, b):

    m = len(y)

    # Calculate linear output
    z = np.dot(X, w) + b

    # Calculate predicted probability
    y_hat = sigmoid(z)

    # Avoid log(0)
    epsilon = 1e-15

    y_hat = np.clip(
        y_hat,
        epsilon,
        1 - epsilon
    )

    # Binary Cross-Entropy Cost
    cost = -(1 / m) * np.sum(
        y * np.log(y_hat)
        +
        (1 - y) * np.log(1 - y_hat)
    )

    return cost


# ------------------------------------------
# 5. Initialize Parameters
# ------------------------------------------

n_features = X_train_scaled.shape[1]

# Initialize weights with zeros
w = np.zeros(n_features)

# Initialize bias with zero
b = 0.0

print("\n========== INITIAL PARAMETERS ==========")

print("Number of features:", n_features)

print("Initial weights:")
print(w)

print("Initial bias:")
print(b)


# ------------------------------------------
# 6. Set Hyperparameters
# ------------------------------------------

learning_rate = 0.01

iterations = 1000

print("\n========== HYPERPARAMETERS ==========")

print("Learning Rate:", learning_rate)

print("Number of Iterations:", iterations)


# ------------------------------------------
# 7. Gradient Descent
# ------------------------------------------

cost_history = []

m = len(y_train_array)

print("\n========== TRAINING MODEL ==========")

for i in range(iterations):

    # --------------------------------------
    # Calculate predicted probabilities
    # --------------------------------------

    z = np.dot(X_train_scaled, w) + b

    y_hat = sigmoid(z)


    # --------------------------------------
    # Calculate Cost
    # --------------------------------------

    cost = compute_cost(
        X_train_scaled,
        y_train_array,
        w,
        b
    )

    cost_history.append(cost)


    # --------------------------------------
    # Calculate Gradients
    # --------------------------------------

    dw = (1 / m) * np.dot(
        X_train_scaled.T,
        (y_hat - y_train_array)
    )

    db = (1 / m) * np.sum(
        y_hat - y_train_array
    )


    # --------------------------------------
    # Update Weights and Bias
    # --------------------------------------

    w = w - learning_rate * dw

    b = b - learning_rate * db


    # --------------------------------------
    # Display Cost
    # --------------------------------------

    if i % 100 == 0:

        print(
            f"Iteration {i}: Cost = {cost:.6f}"
        )


# ------------------------------------------
# 8. Final Cost
# ------------------------------------------

final_cost = cost_history[-1]

print("\n========== TRAINING COMPLETED ==========")

print("Initial Cost:", cost_history[0])

print("Final Cost:", final_cost)


# ------------------------------------------
# 9. Display Final Parameters
# ------------------------------------------

print("\n========== FINAL MODEL PARAMETERS ==========")

print("Final Weights:")
print(w)

print("\nFinal Bias:")
print(b)


# ------------------------------------------
# 10. Plot Cost vs Iteration
# ------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    range(1, iterations + 1),
    cost_history
)

plt.xlabel("Iteration")

plt.ylabel("Cost")

plt.title(
    "Cost vs. Iteration using Gradient Descent"
)

plt.grid(True)

plt.show()


# ------------------------------------------
# 11. Generate Training Probabilities
# ------------------------------------------

train_probabilities = sigmoid(
    np.dot(X_train_scaled, w) + b
)


# ------------------------------------------
# 12. Convert Probabilities to Classes
# ------------------------------------------

train_predictions = (
    train_probabilities >= 0.5
).astype(int)


# ------------------------------------------
# 13. Check Training Accuracy
# ------------------------------------------

training_accuracy = np.mean(
    train_predictions == y_train_array
) * 100

print("\n========== TRAINING RESULT ==========")

print(
    f"Training Accuracy: "
    f"{training_accuracy:.2f}%"
)


# ------------------------------------------
# 14. Display Sample Predictions
# ------------------------------------------

print("\n========== SAMPLE PREDICTIONS ==========")

for i in range(10):

    print(
        f"Sample {i + 1}: "
        f"Actual = {y_train_array[i]}, "
        f"Probability = {train_probabilities[i]:.4f}, "
        f"Predicted = {train_predictions[i]}"
    )


# ------------------------------------------
# 15. Task 7 Summary
# ------------------------------------------

print("\n========== TASK 7 SUMMARY ==========")

print("Logistic Regression implemented from scratch.")

print("Sigmoid function used for probability prediction.")

print("Binary Cross-Entropy used as the cost function.")

print("Gradient Descent used to optimize weights and bias.")

print("Cost history recorded for every iteration.")

print("Cost vs. Iteration plot generated.")

print("\nTask 7 completed successfully!")
```

Precision measures the proportion of transactions predicted as fraudulent that are actually fraudulent.

$$

Recall measures the proportion of actual fraudulent transactions that are correctly identified by the model.

$$
\text{Recall} =
\frac{TP}{TP+FN}
$$

### F1-Score

The F1-Score is the harmonic mean of Precision and Recall.

$$
F1\text{-Score} =
2 \times
\frac{\text{Precision} \times \text{Recall}}
{\text{Precision}+\text{Recall}}
$$
